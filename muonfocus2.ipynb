```python
#evaluates the impact of matter on neutrino oscillations by adjusting the mass squared differences using a matter potential
#which alters the oscillation probabilities for neutrinos traveling through matter.

E_n = 1  # Neutrino energy in GeV
N_e = 1e3   # Electron number density in cm^-3
G_F = 1.1663787e-5  # Fermi coupling constant in GeV^-2


# Mass eigenvalues (in eV^2)
m1_squared = 0.0001
m2_squared = 0.0003
m3_squared = 0.0025
m4_squared = 0.01

# Mixing matrix (for simplicity, assume a diagonal mixing matrix in this example)
U = np.eye(4)

# Matter potential (in GeV)
A = 2 * np.sqrt(2) * G_F * N_e * E_n

# Effective Hamiltonian in matter (diagonal matrix form)
H_eff_matter = np.array([
    [m1_squared + A, 0, 0, 0],
    [0, m2_squared, 0, 0],
    [0, 0, m3_squared, 0],
    [0, 0, 0, m4_squared + A]
])

# Diagonalize the Hamiltonian to get the eigenvalues in matter
eigenvalues_matter = np.linalg.eigvals(H_eff_matter)

# Mass squared differences in vacuum
delta_m21_vacuum = m2_squared - m1_squared
delta_m31_vacuum = m3_squared - m1_squared
delta_m41_vacuum = m4_squared - m1_squared

# Mass squared differences in matter
delta_m21_matter = eigenvalues_matter[1] - eigenvalues_matter[0]
delta_m31_matter = eigenvalues_matter[2] - eigenvalues_matter[0]
delta_m41_matter = eigenvalues_matter[3] - eigenvalues_matter[0]

# Percentage change in mass squared differences due to matter effects
perc_change_m21 = abs((delta_m21_vacuum - delta_m21_matter) / delta_m21_vacuum) * 100
perc_change_m31 = abs((delta_m31_vacuum - delta_m31_matter) / delta_m31_vacuum) * 100
perc_change_m41 = abs((delta_m41_vacuum - delta_m41_matter) / delta_m41_vacuum) * 100

# Output the results
print(f"Mass squared difference (vacuum) Δm21: {delta_m21_vacuum:.6e} eV^2")
print(f"Mass squared difference (matter) Δm21: {delta_m21_matter:.6e} eV^2")
print(f"Percentage change in Δm21 due to matter: {perc_change_m21:.2f}%\n")

print(f"Mass squared difference (vacuum) Δm31: {delta_m31_vacuum:.6e} eV^2")
print(f"Mass squared difference (matter) Δm31: {delta_m31_matter:.6e} eV^2")
print(f"Percentage change in Δm31 due to matter: {perc_change_m31:.2f}%\n")

print(f"Mass squared difference (vacuum) Δm41: {delta_m41_vacuum:.6e} eV^2")
print(f"Mass squared difference (matter) Δm41: {delta_m41_matter:.6e} eV^2")
print(f"Percentage change in Δm41 due to matter: {perc_change_m41:.2f}%")
```

```python
# Constants
Delta_m32_vacuum = 2.4e-3  # eV^2 (vacuum mass squared difference)
A = 2 * np.sqrt(2) * 1.1663787e-5 * 1e3 * 1  # Matter potential factor
L = 1000  # Distance in km

# Matter effect for muon neutrinos
Delta_m32_matter = Delta_m32_vacuum * (1 + A)

# Oscillation probability as a function of energy and distance
def plot_muon_oscillation(E, L):
    P_osc_mu = np.sin(1.267 * Delta_m32_matter * L / E) ** 2  # Oscillation formula
    plt.plot(E, P_osc_mu)
    plt.title('Muon Neutrino Oscillation Probability')
    plt.xlabel('Distance (km)')
    plt.ylabel('Interaction Rate (cm^-2)')
    plt.grid(True)
    plt.show()

# Energy range from 1 to 10 GeV
E = np.linspace(1, 10, 100)

# Plot oscillation probability
plot_muon_oscillation(E, L)
```

```python
# how the charged current interaction rate varies with energy for muon neutrinos
def sigma_CC(E, sin2theta_W=0.23):
    """
    Calculate the charged current interaction cross section
    for muon neutrinos. This is a simplified approximation.
    """
    G_F = 1.1663787e-5  # Fermi coupling constant in GeV^-2
    Q2 = E**2
    cross_section = (G_F**2 * E**2 * (1 - sin2theta_W) * Q2) / (2 * np.pi)

    # Convert from cm^2 to microbarns (1 cm^2 = 1e+24 microbarns)
    cross_section_mb = cross_section * 1e+24  # Convert to microbarns
    return cross_section_mb

E = np.linspace(1, 10, 100)  # Energy range from 1 to 10 GeV
sigma_muon = sigma_CC(E)

# Plotting
plt.plot(E, sigma_muon)
plt.title('Muon Neutrino CC Interaction Cross Section')
plt.xlabel('Energy (GeV)')
plt.ylabel('Cross Section (µb)')  # Updated to microbarns
plt.grid(True)
plt.show()
```

```python
# is meant to look sinusoidal but should have decaying amplitude
Delta_m32 = 2.4e-3  # eV^2
E = 5  # Fixed energy in GeV

# Oscillation probability as a function of distance
def P_osc_mu(L, E):
    return np.sin(1.267 * Delta_m32 * L / E) ** 2

# Effective interaction rate as a function of distance
def effective_interaction_rate(L, E, sin2theta_W=0.23):
    sigma = sigma_CC(E, sin2theta_W)  # Cross-section in microbarns
    return sigma * P_osc_mu(L, E)  # Effective rate

# Distance range (1 km to 10,000 km)
L = np.linspace(1, 10000, 100)
rate_eff = effective_interaction_rate(L, E)

# Plotting
plt.plot(L, rate_eff)
plt.title('Effective Muon Interaction Rate vs. Distance')
plt.xlabel('Distance (km)')
plt.ylabel('Effective Interaction Rate (µb)')
plt.grid(True)
plt.show()
```
