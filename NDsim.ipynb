```python
import numpy as np
import matplotlib.pyplot as plt

# Gaussian distribution parameters (to be adjusted)
mean = 1.8  # Mean energy in GeV
sigma = 0.5  # Standard deviation
num_events = 1000  # Number of events

# Generate Gaussian-distributed random numbers (for simulated neutrino events)
simulated_data = np.random.normal(loc=mean, scale=sigma, size=num_events)

# Plot the histogram of the simulated data
plt.hist(simulated_data, bins=50, density=True, alpha=0.6, color='blue')

# Add labels and show the plot
plt.title('Simulated Gaussian Distribution of Near Detector Events')
plt.xlabel('Energy (GeV)')
plt.ylabel('Density')
plt.show()
```

```python
from scipy.optimize import curve_fit

# Gaussian function for curve fitting
def gaus(x, a, mu, sigma):
    return a * np.exp(-(x - mu)**2 / (2 * sigma**2))

# Fit the Gaussian distribution to your simulated data
bins = np.linspace(min(simulated_data), max(simulated_data), 50)
hist, bin_edges = np.histogram(simulated_data, bins=bins, density=True)
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

# Initial guess for fitting parameters
initial_guess = [1, mean, sigma]
popt, _ = curve_fit(gaus, bin_centers, hist, p0=initial_guess)

# Plot fitted Gaussian
plt.plot(bin_centers, hist, label='Simulated Data', color='blue')
plt.plot(bin_centers, gaus(bin_centers, *popt), label='Gaussian Fit', color='red')
plt.title('Fitted Gaussian to Simulated Data')
plt.xlabel('Energy (GeV)')
plt.ylabel('Density')
plt.legend()
plt.show()

# Output fitted parameters for reference
print(f"Fitted Gaussian parameters: Amplitude = {popt[0]}, Mean = {popt[1]}, Sigma = {popt[2]}")
```
