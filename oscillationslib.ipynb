```python
import numpy as np

# Constants
G_F = 8.9e-5  # Fermi constant in eV * cm^3 / g

def add_matter_effects(hamiltonian, density, electron_fraction=1.0):
    """
    Modify the Hamiltonian to add matter effects.

    Parameters:
    - hamiltonian: 3x3 array, the initial Hamiltonian for oscillations
    - density: float, the matter density in g/cm^3
    - electron_fraction: float, the fraction of electrons in the matter (1 for pure electron density)

    Returns:
    - hamiltonian with matter effects added.
    """
    # Calculate the matter potential
    V = np.sqrt(2) * G_F * density * electron_fraction

    # Add matter effect to the (1,1) position for electron neutrino interactions
    hamiltonian[0, 0] += V
    return hamiltonian

initial_hamiltonian = np.zeros((3, 3))  # Start with a 3x3 Hamiltonian
density = 2.6  # Earth crust average density in g/cm^3
hamiltonian_with_matter = add_matter_effects(initial_hamiltonian, density)
```

```python
def build_3plus1_hamiltonian(theta23, delta_m31, delta_m41, theta34, energy):
    """
    Builds a 4x4 Hamiltonian matrix for a 3+1 neutrino oscillation model.

    Parameters:
    - theta23, theta34: Mixing angles in degrees
    - delta_m31, delta_m41: Mass-squared differences in eV^2
    - energy: Neutrino energy in GeV

    Returns:
    - 4x4 Hamiltonian matrix including sterile neutrino contributions.
    """
    # Convert angles to radians
    theta23_rad = np.radians(theta23)
    theta34_rad = np.radians(theta34)

    # Define the Hamiltonian with delta m^2 terms
    hamiltonian = np.zeros((4, 4))

    # Populate Hamiltonian with active neutrino terms
    hamiltonian[1, 1] = delta_m31 / (2 * energy)
    hamiltonian[2, 2] = delta_m31 / (2 * energy) * np.sin(2 * theta23_rad)

    # Add sterile neutrino terms
    hamiltonian[3, 3] = delta_m41 / (2 * energy)
    hamiltonian[2, 3] = delta_m41 / (2 * energy) * np.sin(2 * theta34_rad)
    hamiltonian[3, 2] = hamiltonian[2, 3]  # Symmetric element

    return hamiltonian
```

```python
theta23 = 45.0
delta_m31 = 2.4e-3  # eV^2
delta_m41 = 1.0     # eV^2 for sterile neutrino
theta34 = 30.0
energy = 1.0  # GeV

# Build and display the Hamiltonian
hamiltonian_3plus1 = build_3plus1_hamiltonian(theta23, delta_m31, delta_m41, theta34, energy)
print("3+1 Hamiltonian:\n", hamiltonian_3plus1)
```
